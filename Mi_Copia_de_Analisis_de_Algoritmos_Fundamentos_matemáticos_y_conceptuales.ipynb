### Uso de las funciones de piso y techo en computación

En el ámbito de las ciencias de la computación, las funciones de **piso** (\\(\lfloor x \rfloor\\) o *floor*, que devuelve el mayor entero menor o igual que \\(x\\)) y de **techo** (\\(\lceil x \rceil\\) o *ceiling*, que devuelve el menor entero mayor o igual que \\(x\\)) son herramientas esenciales para manejar la discretización de variables continuas. Sus aplicaciones principales descritas en los textos incluyen:

1. **Codificación de información y cálculo de bits**:
   La función techo se utiliza directamente para determinar la **cantidad mínima de bits necesarios** para codificar o almacenar de forma única una colección de \\(n\\) objetos, la cual se calcula como \\(\lceil \log_2 n \rceil\\).
2. **Definición de la aritmética modular (operador módulo)**:
   En los sistemas informáticos, el residuo o módulo de una división entera (\\(a \bmod n\\)) se define formalmente en función de la operación de piso:
   \\[a \bmod n = a - n \lfloor a/n \rfloor\\].
3. **Modelado y análisis de ecuaciones de recurrencia**:
   En los algoritmos basados en la estrategia de *divide y vencerás* (como el ordenamiento o la búsqueda), el tamaño de la entrada \\(n\\) frecuentemente se divide entre un factor constante \\(b\\). Dado que \\(n/b\\) no siempre da como resultado un número entero, las relaciones de recurrencia que modelan el tiempo de ejecución se definen asintóticamente utilizando piso y techo (por ejemplo, subdividiendo el problema en tamaños de \\(\lfloor n/b \rfloor\\) o \\(\lceil n/b \rceil\\)) para garantizar que los argumentos de la función de tiempo permanezcan en el dominio de los números enteros.
4. **Acotamiento y discretización en algoritmos**:
   En la implementación y análisis de algoritmos como la **búsqueda binaria**, la localización del elemento central de un intervalo requiere de la división entera de los índices, lo cual equivale matemáticamente a aplicar la función piso. Asimismo, la función techo se utiliza para acotar superiormente el número de términos o el número de pasos en el peor de los casos para ciertos algoritmos, como ocurre con la función de cálculo de potencia recursiva, cuya complejidad en el caso peor está limitada por \\(2 \cdot \lceil \log_2(j) \rceil\\).

### Crecimiento de la función exponencial vs. funciones polinomiales

De acuerdo con el material de estudio, **cualquier función exponencial positiva (con una base estrictamente mayor que 1) crece a un ritmo infinitamente más rápido que cualquier función polinomial**.

La explicación matemática detallada de esta propiedad se fundamenta en el análisis del comportamiento asintótico cuando el tamaño de la entrada \\(n\\) tiende al infinito (\\(n \to \infty\\)). Para cualquier constante exponencial \\(a > 1\\) y cualquier exponente polinomial constante \\(b\\), el límite del cociente de ambas funciones cumple la siguiente condición:

\\[\lim_{n \to \infty} \frac{n^b}{a^n} = 0\\]

A partir de este límite, se deduce que:
*   Un término polinomial está dominado asintóticamente por el exponencial, lo que se denota formalmente en notación asintótica de "o pequeña" como \\(n^b = o(a^n)\\).
*   Como cualquier polinomio \\(p(n)\\) de grado \\(d\\) está asintóticamente acotado y definido por su término de mayor potencia (es decir, \\(p(n) = \Theta(n^d)\\)), el crecimiento del polinomio completo siempre estará subyugado a la tasa de crecimiento del término de mayor orden \\(n^d\\).
*   Dado que cada término \\(n^d\\) individual crece de manera infinitamente más lenta en comparación con la función exponencial \\(a^n\\) (o \\(2^n\\)), un algoritmo con complejidad exponencial multiplicará dramáticamente su tiempo de ejecución ante incrementos mínimos en el tamaño de su entrada \\(n\\), haciéndose impracticable para problemas que no sean extremadamente pequeños.

En términos prácticos, esta diferencia es tan drástica que, mientras que para un tamaño de entrada de \\(n = 100\\), un algoritmo polinomial cuadrático de \\(n^2\\) requiere de \\(10,000\\) operaciones tolerables, uno exponencial de \\(2^n\\) exigiría aproximadamente \\(1.26 \times 10^{30}\\) operaciones; una cifra astronómica que requeriría miles de millones de años de cómputo ininterrumpido incluso en las supercomputadoras más veloces del planeta.

### Concepto de Algoritmo y sus Características Clave

Un **algoritmo** se define formalmente como un **conjunto finito de instrucciones precisas, concretas y no ambiguas** que, al ser ejecutadas en un orden lógico, permiten resolver un problema o calcular una función matemática transformando un conjunto de datos de entrada en una salida esperada. Es, en esencia, un método o proceso sistemático y ordenado diseñado para realizarse de forma mecánica para obtener un resultado en un tiempo acotado.

Para que un procedimiento sea considerado formalmente un algoritmo, debe cumplir con las siguientes **características clave**:

*   **Entrada:** Debe definir de manera precisa los datos o variables que necesita recibir para iniciar su operación.
*   **Salida:** Debe especificar claramente el resultado o lo que produce el algoritmo al concluir.
*   **No ambiguo / Preciso:** Cada instrucción debe ser explícita, clara y detallar sin ambigüedades el orden exacto y la manera en la que deben realizarse los pasos.
*   **Finito:** Debe estar compuesto por una secuencia finita de pasos y **garantizar que la ejecución terminará** tras completar un número limitado de operaciones para cualquier caso de entrada.
*   **Determinista:** Si se ejecuta varias veces el mismo algoritmo sobre el mismo conjunto de datos de entrada, siempre debe pasar por los mismos estados y **producir exactamente los mismos resultados** de salida.
*   **Efectividad:** Cada paso o acción individual debe ser lo suficientemente básico como para poder completarse de manera realizable en un **tiempo finito**.
*   **General:** Debe ser capaz de contemplar y resolver correctamente todos los casos de entrada contemplados por el problema.

### El "Problema de Paro" (Halting Problem) de Alan Turing

Formulado por el matemático Alan Turing en 1936, el **Problema de Paro** consiste en determinar de forma general si cualquier algoritmo (o programa de computadora) dado, al trabajar con una entrada específica, llegará en algún momento a detenerse (es decir, finalizar su ejecución en un tiempo finito) o si continuará ejecutándose indefinidamente en un ciclo infinito.

Turing demostró de manera matemática la **insolubilidad** de este problema, probando que es un **problema indecidible**. Esto significa que **no existe, ni puede existir, un algoritmo general** capaz de examinar cualquier programa y su entrada para decidir correctamente en todos los casos si el programa se detendrá o no. Este hallazgo estableció límites fundamentales e insalvables a lo que la computación y la lógica matemática pueden resolver de forma automatizada.

### Análisis a Priori vs. Prueba a Posteriori para Medir la Eficiencia

La medición de la eficiencia de un algoritmo (tanto en tiempo de ejecución como en espacio de memoria) se divide en dos fases o estudios diferenciados:

| Criterio | **Análisis a Priori (Teórico)** | **Prueba a Posteriori (Empírico)** |
| :--- | :--- | :--- |
| **Definición** | Es un estudio analítico que proporciona una **función matemática** que acota el tiempo de cálculo o los recursos del algoritmo en función del tamaño de los datos de entrada. | Consiste en la recopilación de **estadísticas de tiempo y espacio físicas** consumidas por el algoritmo mientras se ejecuta en un entorno real. |
| **Momento de Aplicación** | Se realiza durante la **etapa de diseño** del algoritmo, mucho antes de que sea codificado e implementado. | Se lleva a cabo en la fase de **verificación, depuración y prueba**, cuando el algoritmo ya está implementado en un programa. |
| **Dependencia del Entorno** | Es **totalmente independiente** del hardware, del lenguaje de programación, del compilador o de la habilidad del programador. | Es **completamente dependiente** de factores coyunturales: requiere una máquina concreta, un compilador específico, un lenguaje definido y datos de prueba particulares. |
| **Propósito Principal** | Permite **predecir asintóticamente el comportamiento del algoritmo** para datos extremadamente grandes, lo que ayuda a evitar el esfuerzo de implementar métodos ineficientes. | Permite evaluar el **rendimiento real bajo condiciones de laboratorio**, recogiendo estadísticas físicas exactas de ejecuciones particulares para optimizar sistemas concretos. |

Ambas metodologías son complementarias; el análisis a priori proporciona las cotas generales que guían la selección del mejor diseño abstracto, mientras que la prueba a posteriori ayuda a los desarrolladores a verificar el comportamiento físico del programa terminado en la realidad computacional.



- *Explica en qué consisten el peor caso, mejor caso y caso promedio al momento de evaluar el comportamiento de un algoritmo ante distintas entradas.*  
**RESPUESTA:**  
En la evaluación del comportamiento de un algoritmo, para un tamaño de entrada fijo **\\(n\\)**, la cantidad de recursos (como el tiempo de ejecución o las operaciones elementales) puede variar drásticamente según las características particulares o la disposición inicial de los datos. Por esta razón, en lugar de un único valor de rendimiento, se analiza un rango de comportamiento acotado por un coste mínimo y un máximo, lo que da lugar a tres escenarios de análisis fundamentales:

### 1. El Peor Caso (*Worst Case*)
*   **Definición:** Es la situación en la cual el algoritmo consume la **máxima cantidad de tiempo, instrucciones o recursos** para resolver un problema de un tamaño \\(n\\) dado. Representa formalmente el límite superior del esfuerzo, es decir, el número máximo de operaciones básicas que el algoritmo puede llegar a ejecutar sobre cualquier entrada posible de tamaño \\(n\\).
*   **Importancia:** Su análisis proporciona una **cota superior** (*upper bound*) para el tiempo de ejecución del algoritmo. Esto ofrece una garantía de que el algoritmo nunca tardará más que ese límite, sin importar qué tan desfavorables sean los datos de entrada. Por este enfoque conservador y seguro, es el análisis más común y prioritario en computación.
*   **Ejemplos de las fuentes:** En una búsqueda secuencial, el peor caso ocurre cuando el elemento buscado se encuentra en la última posición del arreglo o directamente no existe en él (lo que obliga a recorrerlo por completo). En el algoritmo *Quicksort*, el peor caso se da cuando el pivote elegido es el elemento más grande o más pequeño en cada paso, generando subdivisiones extremadamente desequilibradas y degradando el rendimiento a un coste cuadrático \\(O(n^2)\\).

### 2. El Mejor Caso (*Best Case*)
*   **Definición:** Es el escenario en el cual el algoritmo consume la **menor cantidad de tiempo o realiza el mínimo número de operaciones** para resolver una entrada de tamaño \\(n\\). Representa la situación donde el algoritmo se ejecuta de la forma más rápida y conveniente posible.
*   **Importancia:** Establece la cota inferior física del rango de ejecución del algoritmo (el extremo de coste mínimo \\(T_{min}\\)). No obstante, suele considerarse un análisis **"engañoso"** o poco representativo para el diseño práctico de sistemas, ya que es sumamente raro que las condiciones reales de uso coincidan de manera constante con este escenario idílico.
*   **Ejemplos de las fuentes:** En una búsqueda secuencial, el mejor caso ocurre cuando el elemento buscado coincide exactamente con la primera posición del arreglo (requiriendo una única comparación básica). En los algoritmos de ordenamiento clásicos, suele darse cuando el arreglo ya se encuentra perfectamente ordenado desde el inicio.

### 3. El Caso Promedio (*Average Case*)
*   **Definición:** Es el escenario que busca describir el comportamiento del algoritmo ante una **entrada "típica" o aleatoria** de tamaño \\(n\\).
*   **Cálculo e Importancia:** Se considera la **medida más realista** del rendimiento de un algoritmo en entornos de ejecución cotidianos. Sin embargo, su análisis matemático es significativamente más complejo que el del peor o mejor caso. Para calcularlo, es obligatorio establecer o asumir una **distribución de probabilidad estadística** (\\(Pr(I)\\)) sobre la ocurrencia de cada entrada de tamaño \\(n\\), ponderando el coste de cada una por su probabilidad individual:
    \\[A(n) = \sum_{I \in D_n} Pr(I) \cdot t(I)\\]
    Dado que en la práctica es difícil determinar analíticamente esta distribución real, los desarrolladores suelen realizar supuestos simplificadores, como asumir que todas las entradas de tamaño \\(n\\) son igualmente probables (distribución uniforme).
*   **Ejemplos de las fuentes:** El algoritmo *Quicksort* es un ejemplo clásico de la importancia de este análisis: mientras que su peor caso es cuadrático, su rendimiento promedio es de \\(O(n \log n)\\) comparaciones. Esto demuestra que los casos malos ocurren con tan poca frecuencia en permutaciones aleatorias que el algoritmo sigue siendo uno de los métodos de ordenamiento más rápidos y preferidos en la práctica.

1. Implementa una versión recursiva para el cálculo del factorial ($n! = n \cdot (n-1)!$) y una versión iterativa para Fibonacci.

````python
# 1. Versión recursiva para el cálculo del factorial (n! = n * (n - 1)!)
def factorial_recursivo(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursivo(n - 1)

# Versión iterativa para Fibonacci
def fibonacci_iterativo(n):
    if n <= 1:
        return n
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

# Pruebas rápidas
print("Factorial de 5 (recursivo):", factorial_recursivo(5))
print("Fibonacci de 10 (iterativo):", fibonacci_iterativo(10))
````

2. Utiliza la librería `time` o la función mágica `%timeit` para comparar el tiempo consumido (A posteriori) entre el Fibonacci recursivo y el iterativo para `n=35`.

````python
import time

# Definimos también el Fibonacci recursivo para poder compararlo
def fibonacci_recursivo(n):
    if n <= 1:
        return n
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)

n = 35

# Medición A posteriori: Fibonacci Recursivo
inicio_rec = time.perf_counter()
res_rec = fibonacci_recursivo(n)
fin_rec = time.perf_counter()
tiempo_rec = fin_rec - inicio_rec

# Medición A posteriori: Fibonacci Iterativo
inicio_ite = time.perf_counter()
res_ite = fibonacci_iterativo(n)
fin_ite = time.perf_counter()
tiempo_ite = fin_ite - inicio_ite

print(f"Fibonacci Recursivo (n={n}) -> Resultado: {res_rec} | Tiempo: {tiempo_rec:.6f} segundos")
print(f"Fibonacci Iterativo (n={n}) -> Resultado: {res_ite} | Tiempo: {tiempo_ite:.6f} segundos")
print(f"Diferencia: La versión iterativa fue {tiempo_rec / tiempo_ite:.0f} veces más rápida.")
````

| Criterio de Análisis | Fibonacci Iterativo | Fibonacci Recursivo |
| :--- | :--- | :--- |
| **Complejidad Temporal (A priori)** | $O(n)$ (Lineal / Polinomial de grado 1) | $O(2^n)$ (Exponencial) |
| **Complejidad Espacial (Memoria)** | $O(1)$ (Constante: solo usa dos variables auxiliares `a` y `b`) | $O(n)$ (Lineal: acumula llamadas en la pila de ejecución) |
| **Comportamiento para $n = 35$** | Ejecución prácticamente instantánea (microsegundos) | Tarda varios segundos debido a millones de llamadas redundantes |
| **Crecimiento de operaciones** | Realiza exactamente $n - 1$ iteraciones directas | Duplica las llamadas en cada nivel del árbol de recursión |

**Conclusión:**
La versión **iterativa** crece de forma lineal/polinomial ($O(n)$) porque utiliza un único ciclo `for` que avanza secuencialmente desde $2$ hasta $n$, almacenando y reutilizando los dos resultados anteriores en memoria sin repetir cálculos. En cambio, la versión **recursiva** tiene un comportamiento exponencial ($O(2^n)$) porque en cada llamada se bifurca en dos subproblemas nuevos ($F(n-1)$ y $F(n-2)$) sin guardar resultados previos, recalculando los mismos subproblemas millones de veces conforme crece $n$.
